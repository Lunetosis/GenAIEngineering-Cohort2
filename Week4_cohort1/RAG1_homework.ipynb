# Assignment: Building a Basic AI-Powered Search for Data Dynamics (Based on Sessions 1 & 2)

**Scenario**: 

You are part of the Data Dynamics team (Anya, Ben, Clara). Your task is to take the first concrete step towards solving their internal knowledge access problem. You need to build a basic AI-powered search engine that can answer questions based on a small collection of internal documents (provided or chosen by you, e.g., a few internal reports or technical notes).

**Goal**:

Implement a system that can ingest unstructured text data, process it for search, and retrieve relevant information snippets in response to a user query. This directly applies the foundational concepts from Sessions 1 and 2.

## Conceptual Coding Activity / Assignment Steps:

    1. Data Loading & Preparation:

        - Load a small set of text documents (e.g., 3-5 documents). You can use standard Python file reading or a library if needed (though loaders weren't explicitly detailed in Session 1, assume basic text files are provided).

        - Implement document chunking. Choose either Fixed-Size Chunking using CharacterTextSplitter or Recursive Character Text Splitting using RecursiveCharacterTextSplitter. Experiment with chunk_size and chunk_overlap. Explain why you chose your specific chunking strategy and parameters, referencing the trade-offs discussed in Session 1. Print or inspect the first few chunks to verify your splitting works as expected.

    2. Vectorization: Creating Embeddings:

        - Select a Sentence Transformer model, such as the recommended all-MiniLM-L6-v2.
        - Using the sentence-transformers library or a LangChain wrapper like HuggingFaceEmbeddings, generate vector embeddings for each of the text chunks created in the previous step.
        -Note the dimensionality of the generated embeddings.


    3. Storing Embeddings: Using a Vector Database:

        - Initialize a vector database using either FAISS or Qdrant in in-memory mode. Explain why you chose this specific vector store, considering the points discussed about their suitability for local development.
        - Add the generated vector embeddings to your chosen vector store. Associate each vector with its corresponding text chunk and potentially simple metadata (like a document identifier) if using Qdrant.

    4. Implementing Retrieval (Semantic Search):

        - Implement a semantic search function. This function should take a user query as input.
        - Use the same embedding model as before to generate an embedding for the user query.
        - Use the vector store's search capability (e.g., similarity_search in LangChain wrappers) to find the top K (e.g., K=3 or 5) most similar vectors to the query vector.
        - Retrieve the text chunks associated with these most similar vectors.

    5. Presenting Results:
        - For a given sample query, run your semantic search function.
        - Print the original user query and the text content of the top K retrieved chunks. This simulates presenting the "relevant information" to the user.

## Optional (Bonus):
    - Implement a simple Keyword Search (BM25) retrieval function using a library like rank_bm25. Explain when keyword search might be more effective than semantic search for certain types of queries.
    - Wrap your search functionality in a basic command-line interface or a simple web UI using Streamlit/Gradio (drawing on prior bootcamp knowledge).
