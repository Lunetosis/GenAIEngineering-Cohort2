1. Write a system prompt that gets you exact answers/numbers from annual report of a company.

    a. What other parameters would you like to control

    b. Sample report: https://s201.q4cdn.com/141608511/files/doc_financials/2025/annual/NVIDIA-2025-Annual-Report.pdf

```python
#Prompt to extract info from report

company_name = "NVIDIA"
report_url = "https://s201.q4cdn.com/141608511/files/doc_financials/2025/annual/NVIDIA-2025-Annual-Report.pdf"
year = 2025
financial_metrics = [
    "total revenue",
    "net income",
    "earnings per share (EPS)",
    "operating expenses",
    "gross profit",
    "research and development expenses",
    "cash and cash equivalents",
    "total assets",
    "total liabilities",
    "tier3 capital ratio",
    "return on equity (ROE)",
    "return on assets (ROA)",
    "debt to equity ratio",
    "current ratio",
    "quick ratio",
    "dividend yield",
    "price to earnings (P/E) ratio",
    "price to book (P/B) ratio",
    "market capitalization",
    "free cash flow",
    "operating cash flow",
    "capital expenditures (CapEx)",
    "working capital",
    "inventory turnover",
    "accounts receivable turnover",
    "accounts payable turnover",
    "net profit margin",
    "operating profit margin",
    "gross profit margin",
    "earnings before interest and taxes (EBIT)",
    "earnings before interest, taxes, depreciation, and amortization (EBITDA)",
    "segment performance",
    "geographic revenue breakdown"]

prompt = f"You are an experienced financial analyst. Your task is to extract specific financial information from the annual report of {company_name}, for the year of {year}. The report is structured with sections and subsections, and you need to provide precise answers based on the content."

prompt += f"\n\nPlease extract the following financial metrics from the report:\n"
prompt += "\n".join(financial_metrics)
prompt += f"\n\nThe report can be found at: {report_url}\n"
prompt += "Please provide the answers in a structured format, with each metric clearly labeled and its value provided. If a metric is not found in the report, indicate that it is not available." 
prompt += "\n\nEnsure that the values are accurate and reflect the latest financial data presented in the report. If there are multiple values for a metric, provide the most relevant one based on the context of the report."
prompt += "\n\nIf you encounter any issues or uncertainties while extracting the information, please note them clearly in your response."
prompt += "\n\nYour response should be concise, clear, and focused solely on the financial metrics requested. Avoid any unnecessary commentary or explanations."
prompt += "\n\nMake sure to present references to where the information is extracted from and provide links."
print(prompt)
```
